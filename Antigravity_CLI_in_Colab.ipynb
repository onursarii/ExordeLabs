# Antigravity CLI in Colab

The Antigravity CLI (`agy`) is a coding agent that runs on your Colab runtime, next to the kernel. You chat with it in Colab's terminal, and it adds, edits, and runs cells in this notebook. Because it runs on the runtime, it can also use the shell to read files, install packages, check the GPU, and run scripts.

This notebook shows how to:

1. Open `agy` and sign in.
2. Ask `agy` to plot data, fix a broken cell, and run a script on the runtime.
3. Install and write skills, and choose a model and mode.
4. Optionally, sign in with a Google Cloud project.

## Get started

To open the Antigravity CLI, press `Ctrl+Shift+P` (`Cmd+Shift+P` on macOS) to open the command palette, then select **Antigravity CLI**. You can also click **Terminal** at the bottom of the page and run `agy`.

The first time you run `agy` on a runtime, follow the prompts to sign in. Colab saves your `agy` settings to your account, so you only choose them once. To sign in with a Google Cloud project instead, see [Optional: Sign in with a Google Cloud project](#scrollTo=agyAdcIntro01).

## Examples

Each example has a prompt with a copy icon. Click the icon, paste the prompt into the `agy` chat, and press Enter. `agy` changes the cells under the example, so keep this notebook open to see its work.

### Example 1: Plot data

`agy` can read files on the runtime and write code that uses them. This prompt asks it to plot the California housing dataset from Colab's `sample_data` folder. `agy` adds a code cell under this example and runs it.

```python
# @markdown *Paste this prompt into the `agy` chat.*
import html

from IPython.display import HTML, display


def show_prompt(*prompts):
  """Shows each prompt in a box with a copy icon."""
  script = """<script>
function copyPrompt(button) {
  const area = document.body.appendChild(document.createElement('textarea'));
  area.value = button.previousElementSibling.textContent;
  area.select();
  document.execCommand('copy');
  area.remove();
  button.lastChild.hidden = false;
  setTimeout(() => button.lastChild.hidden = true, 2000);
}
</script>"""
  box = """<div style="display:flex;gap:8px;margin:0 0 12px;padding:10px 12px;border:1px solid #c4c7c5;border-radius:8px">
<pre style="flex:1;white-space:pre-wrap;margin:0;font-size:13px;line-height:1.5">{}</pre>
<button onclick="copyPrompt(this)" title="Copy" aria-label="Copy prompt" style="cursor:pointer;border:none;background:none;color:inherit;font-size:12px"><svg viewBox="0 0 24 24" width="18" height="18" fill="currentColor"><path d="M16 1H4c-1.1 0-2 .9-2 2v14h2V3h12V1zm3 4H8c-1.1 0-2 .9-2 2v14c0 1.1.9 2 2 2h11c1.1 0 2-.9 2-2V7c0-1.1-.9-2-2-2zm0 16H8V7h11v14z"/></svg><span hidden> Copied</span></button>
</div>"""
  display(HTML(script + ''.join(box.format(html.escape(p)) for p in prompts)))


show_prompt(
    'Read @sample_data/california_housing_train.csv and add a code cell under'
    ' Example 1 that plots a scatter plot of longitude vs latitude, colored by'
    ' median_house_value.',
)
```

### Example 3: Use the runtime

`agy` isn't limited to notebook cells. It can also use the runtime's shell, files, and background processes.

Paste these prompts one at a time. The first asks about the runtime's hardware. The second writes a training script, runs it as a separate process, and tells you the test score in the chat. The kernel stays free, so you can keep running cells while the model trains.

```python
# @markdown *Paste these prompts into the `agy` chat, one at a time.*
show_prompt(
    'What hardware does this runtime have (CPU, RAM, GPU, and free disk)? Add a'
    ' markdown cell below Example 3 with a short table.',
    'Write a script at /content/train.py that trains a gradient boosting'
    ' regressor on @sample_data/california_housing_train.csv, evaluates it on'
    ' @sample_data/california_housing_test.csv, and logs progress to'
    ' /content/train.log. Run it in the background and tell me the test score'
    ' when it finishes.',
)
```

## Customize agy

### Install a skill from GitHub

A skill is a folder with a `SKILL.md` file that teaches `agy` a workflow. The [google/skills](https://github.com/google/skills) repository has skills for Google products.

The following cell installs `finding-google-skills`, which helps `agy` find the Google skills and products that fit a task. To install a different skill, paste the URL of its GitHub folder into `skill_url`. If `agy` is already running, type `/skills reload` in the chat to load the new skill.

```python
# @title Install a skill from GitHub
skill_url = "https://github.com/google/skills/tree/main/skills/developers/finding-google-skills"  # @param {type:"string"}

repo, _, rest = skill_url.partition("/tree/")
branch, _, path = rest.partition("/")
!rm -rf /tmp/skill-repo
!git clone --quiet --depth 1 --branch {branch} {repo} /tmp/skill-repo
!mkdir -p ~/.gemini/config/skills && cp -r /tmp/skill-repo/{path} ~/.gemini/config/skills/
print('Installed. If agy is already running, type "/skills reload" in agy.')
```

To try the skill, paste the following prompt. `agy` searches Google's skill catalog and adds a markdown cell with the matching skill and its GitHub folder URL. To install that skill too, paste the URL into `skill_url` and run the install cell again.

```python
# @markdown *Paste this prompt into the `agy` chat.*
show_prompt(
    '/finding-google-skills Which Google skill helps me work with Cloud'
    ' Storage? Add a markdown cell in the "Install a skill from GitHub" section'
    ' of the notebook with its name and GitHub folder URL.'
)
```

### Write a skill

You can also write your own skills. The following cell creates a `publication-charts` skill that formats the charts `agy` adds to your notebook for reports and slides. To read or edit the skill, expand the cell. If `agy` is already running, type `/skills reload` after you run the cell.

```python
# @title Create the publication-charts skill
from pathlib import Path

skill_dir = Path.home() / '.gemini' / 'config' / 'skills' / 'publication-charts'
skill_dir.mkdir(parents=True, exist_ok=True)
(skill_dir / 'SKILL.md').write_text("""---
name: publication-charts
description: >-
  Makes the charts added to the open Colab notebook ready for a report or
  slides: a title that states the takeaway, labeled axes, colorblind-safe
  colors, and a saved PNG. Use when asked to plot, chart, or visualize data.
---

# Publication-ready charts

When you add a chart cell to the notebook:

1. Use matplotlib, or seaborn for statistical plots, with
   `plt.style.use('tableau-colorblind10')`.
2. Write a title that states the takeaway, such as "Prices rise with income",
   rather than naming the axes.
3. Label both axes with units, and format large numbers with thousands
   separators.
4. Use `figsize=(10, 6)`, font sizes of at least 12, and remove the top and
   right spines.
5. Create `/content/figures` if it doesn't exist, then save the figure to
   `/content/figures/<short-name>.png` at `dpi=200` before showing it, and
   mention the path in your reply.
""")
print(f'Wrote {skill_dir / "SKILL.md"}')
print(
    'If agy is already running, type "/skills reload" in agy to load the skill.'
)
```

To try the skill, paste the following prompt. Typing `/publication-charts` before your request runs that skill. `agy` also uses a skill without the slash command when your request matches the skill's description. The chart is saved to `/content/figures`.

```python
# @markdown *Paste this prompt into the `agy` chat.*
show_prompt(
    '/publication-charts how median house value changes with median income in'
    ' @sample_data/california_housing_train.csv.'
)
```

### More ways to customize

- **Model and mode:** type `/model` to choose a model. A Flash model is faster for small changes. Press `Shift+Tab` to cycle through agent modes: in **accept-edits** mode, `agy` edits files without asking, and in **plan** mode, it researches and plans without making changes. To set a default mode, go to **Agent Mode** in `/config`.
- **Plugins:** type `/plugin` to browse the Antigravity marketplace for bundles of skills, MCP servers, and subagents.
- **Rules and MCP servers:** put standing instructions in `~/.gemini/GEMINI.md`. To add an MCP server, run `agy mcp add` in a terminal or edit `~/.gemini/config/mcp_config.json`. Type `/mcp` to see your servers.

Colab saves your `agy` settings, including the default mode, to your account. Files you add under `~/.gemini`, such as skills and rules, are kept only for the current runtime.

## Optional: Sign in with a Google Cloud project

If you have a Google Cloud project, `agy` can sign in with Application Default Credentials and use the project's quota instead of your personal sign-in. Enter your project ID in the following cell and run it. The cell signs you in and sets the environment variables that `agy` reads when it starts in a new terminal session.

Then restart `agy` in the terminal:

1. Type `/exit` to quit `agy`.
2. Run `source ~/.bashrc`.
3. Run `agy`.

```python
# @title Sign in with Application Default Credentials
GCP_PROJECT = ''  # @param {type:"string"}

import os
import re

from google.colab import auth

if not GCP_PROJECT:
  raise ValueError('Enter your Google Cloud project ID.')

auth.authenticate_user()

name = 'agy-adc'
block = (
    'unset GEMINI_API_KEY\n'
    'export AGY_ADC_AUTH=true\n'
    f'export GOOGLE_CLOUD_PROJECT={GCP_PROJECT}\n'
    f'export GOOGLE_CLOUD_QUOTA_PROJECT={GCP_PROJECT}\n'
)
with open(f'/etc/profile.d/{name}.sh', 'w') as f:
  f.write(block)

bashrc = os.path.expanduser('~/.bashrc')
begin, end = f'# >>> {name} >>>', f'# <<< {name} <<<'
rc = open(bashrc).read() if os.path.exists(bashrc) else ''
rc = re.sub(rf'{re.escape(begin)}.*?{re.escape(end)}\n', '', rc, flags=re.S)
with open(bashrc, 'w') as f:
  f.write(f'{rc.rstrip()}\n{begin}\n{block}{end}\n')

print(f'Signed in. agy uses project {GCP_PROJECT} in new terminal sessions.')
```
